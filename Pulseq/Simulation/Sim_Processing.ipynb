# Sim Data Processing

```python
import sys
import h5py
import numpy as np
import matplotlib.pyplot as plt

%matplotlib qt
```

```python
def load_sim(path):
    with h5py.File(path, "r") as f:
        signal = f["signal_real"][()] + 1j * f["signal_imag"][()]   # (Nsamples, Ncoils)
        out = {
            "signal": signal,
            "t_adc": f["t_adc"][()] if "t_adc" in f else None,           # seconds
            "kspace": f["kspace_adc"][()] if "kspace_adc" in f else None,  # (Nsamples, >=3), 1/m
            "attrs": dict(f.attrs),
        }
    return out
```

```python
sim = load_sim('sim_output.h5')
# signal = sim['signal'].reshape(64,2,5,10)
signal = sim['signal'].reshape(5,2,10,64)

sft = np.fft.fftshift(np.fft.fft(np.fft.fftshift(signal, axes=(3)), axis=3), axes=(3))
plt.figure()
# plt.plot(np.abs(sft[0,0,0,:]))
for i in range(5):
    plt.plot(np.concatenate(np.abs(sft[i,0,:,:] + sft[i,1,:,:])))
plt.show()
```
